**Q14:- To write a Python program to reverse the digits of a given number.**

```python
num = int(input("Enter a number: "))
rev = 0
while num>0:
  digit = num%10
  rev = rev*10+digit
  num //=10
print("Reversed number: ",rev)
```

**Q17:- To write a Python program to check whether a given number is prime.**

```python
num = int(input("Enter a number: "))
if num>1:
  for i in range(2,num):
    if num % i == 0:
      print("Not Prime")
      break
  else:
      print("Number is Prime")
else:
    print("Not Prime")
```

**Q18:- To write a Python program to generate the Fibonacci series up to N terms.**

```python
n = int(input("Enter number of terms: "))
a,b=0,1
for i in range(n):
  print(a, end=" ")
  a, b = b, a+b
```

**Q19:- To write a Python program to check whether a string is a palindrome.**

**Q20:- To write a Python program to check whether a given number is an Armstrong number.**

```python
num = int(input("Enter a number: "))
sum = 0
temp = num
while temp > 0:
    digit = temp % 10
    sum += digit ** 3
    temp //= 10
if sum == num:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```
